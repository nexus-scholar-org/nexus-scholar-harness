This notebook demonstrates how to interactively define an `IntentPacket`, compile it into a canonical `ResearchProtocol`, compute its deterministic fingerprint, and render human-readable PRISMA screening criteria.

```python
import json
from pathlib import Path
from scholar_protocol.compiler import compile_protocol
from scholar_protocol.intent import IntentPacket
from scholar_protocol.canonical import canonical_json, canonical_fingerprint
from scholar_protocol.render import render_screening_criteria
from scholar_protocol.validate import validate_protocol
```

## 1. Define Research Intent Profile

```python
intent_data = {
    "protocol_id": "proto-2026-notebook-demo",
    "genesis_timestamp": "2026-09-01T00:00:00+00:00",
    "project_slug": "transformer-attention-benchmark",
    "playbook_type": "DESIGN_SCIENCE",
    "title": "Empirical Efficiency of Sub-Quadratic Attention Mechanisms",
    "lead_researcher": "Dr. Scholar",
    "unit_of_analysis": "Transformer Attention Mechanisms",
    "epistemological_rationale": "Controlled quantitative benchmark of sequence throughput and memory complexity.",
    "research_questions": [
        {
            "text": "What computational throughput and memory scaling improvements are achieved by sub-quadratic attention?",
            "target_facet": "evaluation_metrics",
            "required_evidence_type": "Quantitative Benchmark"
        }
    ],
    "core_concepts": [
        {"concept": "Linear Attention", "synonyms": ["kernelized attention", "sub-quadratic attention"]}
    ],
    "inclusion_criteria": [
        {"criterion": "Reports empirical throughput, memory, or latency benchmarks on language/vision tasks", "maps_to_rqs": ["RQ1"]}
    ],
    "exclusion_criteria": [
        {"criterion": "Pure theoretical proposals without empirical benchmark data", "reason_category": "METHODOLOGY", "maps_to_rqs": ["RQ1"]}
    ],
    "matrix_dimensions": [
        {"id": "scaling_complexity", "name": "Scaling Complexity", "description": "Theoretical asymptotic complexity O(N) or O(N sqrt(N))"},
        {"id": "throughput_tokens_sec", "name": "Throughput", "description": "Reported tokens/sec throughput during inference"}
    ]
}

intent = IntentPacket.model_validate(intent_data)
print(f"Intent loaded for project: {intent.project_slug}")
```

## 2. Compile into Canonical Research Protocol

```python
protocol = compile_protocol(intent)
fingerprint = canonical_fingerprint(protocol)
print(f"Canonical Protocol Fingerprint: {fingerprint}")

# Save protocol.json
proto_path = Path("protocol.json")
proto_path.write_bytes(canonical_json(protocol))
print(f"Saved canonical protocol to {proto_path}")
```

## 3. Render Human-Readable Screening Criteria

```python
criteria_md = render_screening_criteria(protocol)
from IPython.display import display, Markdown
display(Markdown(criteria_md))
```
